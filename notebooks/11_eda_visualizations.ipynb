# EDA visualizations


Exploratory plots. PNGs also in `figures/`.


```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
DATA, FIG = Path("../data"), Path("../figures")
churn = pd.read_csv(DATA / "churn.csv")
houses = pd.read_csv(DATA / "house_prices.csv")
sales = pd.read_csv(DATA / "daily_sales.csv", parse_dates=["date"])
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
churn["churn"].value_counts().plot(kind="bar", ax=axes[0])
axes[1].hist(churn["tenure_months"], bins=20)
plt.tight_layout(); plt.savefig(FIG / "eda_churn_overview.png", dpi=120); plt.show()
plt.figure(figsize=(6, 4))
plt.scatter(houses["dist_center_km"], houses["price"], alpha=0.3, s=10)
plt.savefig(FIG / "eda_price_vs_distance.png", dpi=120); plt.show()
plt.figure(figsize=(10, 3)); plt.plot(sales["date"], sales["sales"], lw=0.7)
plt.savefig(FIG / "eda_sales_line.png", dpi=120); plt.show()
```
