```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
# ======================
# 1. ПАРАМЕТРЫ И ПУТИ
# ======================
DATA = Path("../data")
FIG = Path("../figures")
FIG.mkdir(parents=True, exist_ok=True)
```

```python
churn = pd.read_csv(DATA / "churn.csv")
houses = pd.read_csv(DATA / "house_prices.csv")
sales = pd.read_csv(DATA / "daily_sales.csv", parse_dates=["date"])
```

```python
print("churn:", churn.shape)
print("houses:", houses.shape)
print("sales:", sales.shape)
```

```python
# ======================
# 2. CHURN: ОБЗОР
# ======================
fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Столбцы churn
churn["churn"].value_counts().sort_index().plot(
    kind="bar", ax=axes[0], color=["#440154", "#21918c"], edgecolor="white"
)
axes[0].set_title("Распределение churn", fontsize=12)
axes[0].set_xlabel("churn")
axes[0].set_ylabel("Количество")
axes[0].grid(axis="y", linestyle="--", alpha=0.5)

# Гистограмма tenure
sns.histplot(churn["tenure_months"], bins=20, ax=axes[1], color="#35b779", edgecolor="white")
axes[1].set_title("Tenure (месяцы)", fontsize=12)
axes[1].set_xlabel("tenure_months")
axes[1].grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig(FIG / "eda_churn_overview.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# Churn vs monthly_charges — violin
plt.figure(figsize=(8, 5))
sns.violinplot(
    data=churn,
    x="churn",
    y="monthly_charges",
    palette="rocket",
    hue="churn",
    legend=False,
    inner="quartile",
)
plt.title("Monthly charges по churn", fontsize=14, pad=15)
plt.xlabel("churn", fontsize=12)
plt.ylabel("monthly_charges", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / "eda_churn_violin.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# ======================
# 3. HOUSES: ЦЕНА VS РАССТОЯНИЕ
# ======================
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=houses,
    x="dist_center_km",
    y="price",
    hue="rooms",
    palette="viridis",
    alpha=0.45,
    s=28,
)
plt.title("Цена vs расстояние до центра", fontsize=14, pad=15)
plt.xlabel("dist_center_km", fontsize=12)
plt.ylabel("price", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / "eda_price_vs_distance.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# Корреляции признаков домов
corr = houses.corr(numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5,
)
plt.title("Корреляции: house_prices", fontsize=14, pad=15)
plt.tight_layout()
plt.savefig(FIG / "eda_houses_corr.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# ======================
# 4. SALES: ВРЕМЕННОЙ РЯД
# ======================
daily = sales.groupby("date", as_index=False)["sales"].sum()

plt.figure(figsize=(12, 3.5))
plt.plot(daily["date"], daily["sales"], lw=0.8, color="#31688e")
plt.title("Суммарные дневные продажи", fontsize=14, pad=12)
plt.xlabel("date", fontsize=12)
plt.ylabel("sales", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / "eda_sales_line.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# Продажи по дням недели
sales = sales.copy()
sales["dayofweek"] = sales["date"].dt.day_name()
order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

plt.figure(figsize=(10, 5))
sns.boxplot(
    data=sales,
    x="dayofweek",
    y="sales",
    order=order,
    palette="viridis",
    hue="dayofweek",
    legend=False,
)
plt.title("Продажи по дням недели", fontsize=14, pad=15)
plt.xlabel("")
plt.ylabel("sales", fontsize=12)
plt.xticks(rotation=30)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / "eda_sales_by_weekday.png", dpi=120, bbox_inches="tight")
plt.show()
```

```python
# Promo effect
plt.figure(figsize=(7, 5))
sns.boxplot(data=sales, x="promo", y="sales", palette="mako", hue="promo", legend=False)
plt.title("Влияние promo на продажи", fontsize=14, pad=15)
plt.xlabel("promo (0/1)", fontsize=12)
plt.ylabel("sales", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.4)
plt.tight_layout()
plt.savefig(FIG / "eda_sales_promo.png", dpi=120, bbox_inches="tight")
plt.show()
```
